```python
def hamming(hash1,hash2):
    L = len(hash1)
    distance = 0
    
    for i in range(L):
        if hash1[i] != hash2[i]:
            distance += 1
    
    print("Hamming Distance is: " + str(distance))
```

```python
def show(hash_type,hash_Value):
    print(str(hash_type) + ": " + str(hash_Value))
```

```python
#Array com todas as Ditâncias de Hamming Identificadas
Original_array = [0,0,0,0,0,0,0]
Ahash_Array = [0,4,2,16,9,5,6]
Phash_Array = [0,2,2,16,13,7,11]
Dhash_Array = [0,7,7,16,12,5,3]
Whash_Array = [0,2,2,16,11,6,7]
Hsvc_Array = [0,0,0,1,1,1,1]
CPhash_Array = [0,15.1,14.4,15.9,15.9,16,15.6]
```

```python
def distance_Calc(array,algoritm):
    dst_euclid = distance.euclidean(Original_array, array)
    dst_manh = distance.cityblock(Original_array, array)
    
    print("\nAlgoritmo: " + str(algoritm))
    print("O Valor da Distância Euclidiana é: " + str(dst_euclid))
    print("O valor da Distância de Manhattan é: " + str(dst_manh))
```

```python
distance_Calc(Ahash_Array, "Average Hashing")
distance_Calc(Phash_Array, "Perceptual Hashing")
distance_Calc(Dhash_Array, "Difference Hashing")
distance_Calc(Whash_Array, "Wavelet Hashing")
distance_Calc(Hsvc_Array, "HSV-Color Hashing")
distance_Calc(CPhash_Array, "Crop-Resistant Hashing")
```
